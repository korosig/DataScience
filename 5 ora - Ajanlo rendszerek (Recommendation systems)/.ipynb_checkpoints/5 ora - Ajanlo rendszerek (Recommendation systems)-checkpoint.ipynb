### Hogyan találhatunk hasonló felhasználókat a termék minősítés alapján?

A hasonlóság fogalmának megértéséhez először hozzunk létre egy egyszerű adathalmazt.

Az adatok négy felhasználó, A, B, C és D , akik két filmet értékeltek. Az értékeléseket listákban tároljuk, és minden lista két számot tartalmaz, amelyek az egyes filmek értékelését jelzik:

````python
a = [1, 2]
b = [2, 4]
c = [2.5, 4]
d = [4.5, 5]
````

Szemléltessük ezt vizuálissan is.

````python
import matplotlib.pyplot as plt
labels = ['a','b','c','d']
for i,point in enumerate([a,b,c,d]):
  plt.scatter(point[0],point[1])
  plt.text(point[0]-0.1,point[1]+0.1,s="user_" + labels[i])
  plt.title('Felhasználók válasza')
  plt.xlabel('1. film')
  plt.ylabel('2. film')
````

Ha a pontok közötti távolságot nézzük, az jó módnak tűnik a hasonlóság becslésére, igaz?

A távolságot a két pont közötti euklideszi távolság képletével határozhatjuk meg. Ehhez a scipy-ben elérhető függvényt használhatjuk, a következő képpen:

````python
from scipy import spatial

a = [1, 2]
b = [2, 4]
c = [2.5, 4]
d = [4.5, 5]
print('Távolság C -> A')
print(spatial.distance.euclidean(c, a))
print('Távolság C -> B')
print(spatial.distance.euclidean(c, b))
print('Távolság C -> D')
print(spatial.distance.euclidean(c, d))
````

Láthatjuk, hogy a C legközelebbi "szomszédja" (hozzá legjobban hasonló) felhasználó a B.

````python
print('Távolság C -> B: ' + str(spatial.distance.euclidean(c, b)))
````

De A és D közül kihez áll közelebb a C?

Mondhatnánk, hogy C közelebb van D-hez a távolságot tekintve.

A vélemények pontjait nézve úgy tűnik, hogy C választása jobban egybeesik A választásával, mint D-é, mert A és C is majdnem kétszer annyira szereti a második filmet, mint az elsőt, a D-nek pedig mindkét film egyformán tetszik

````python
print('vélemény A')
print(a)
print('vélemény B')
print(b)
print('vélemény C')
print(c)
print('vélemény D')
print(d)
````

Mivel tudnánk hatékonyan az olyan mintákat azonosítani, amelyeket az euklideszi távolság nem tud?

Egy jó megoldás ha az origót és az adott pontokat összekötő vonalak közötti szöget(Cos) figyeljük meg.

````python
import matplotlib.pyplot as plt
labels = ['a','b','c','d']
for i,point in enumerate([a,b,c,d]):
  plt.plot([0,point[0]],[0,point[1]])
  plt.scatter(point[0],point[1])
  plt.text(point[0]-0.1,point[1]+0.1,s="user_" + labels[i])
  plt.title('Felhasználók válasza')
  plt.xlabel('1. film')
  plt.ylabel('2. film')
````

A grafikon négy egyenest mutat, amelyek minden egyes pontot az origóhoz kötnek. Az A és B egyenesek egybeesnek, így a köztük lévő szög nulla.

A hasonlóság kiszámításához a szög segítségével olyan függvényre van szükség, amely kisebb szög esetén nagyobb hasonlóságot vagy kisebb távolságot, nagyobb szög esetén pedig kisebb hasonlóságot vagy nagyobb távolságot ad vissza. A szög koszinusza egy olyan függvény, amely 1-től -1-ig csökken, ahogy a szög 0-tól 180-ig növekszik.

````python
print('Koszinusz távolság C -> A')
print(spatial.distance.cosine(c,a))
print('Koszinusz távolság C -> B')
print(spatial.distance.cosine(c,b))
print('Koszinusz távolság C -> D')
print(spatial.distance.cosine(c,d))
````

A C és A vektorok közötti kisebb szög, tehát alacsonyabb a koszinusz távolság.

````python
print('Koszinusz távolság C -> A')
print(spatial.distance.cosine(c,a))
````

Vegyük észre, hogy A és B felhasználó a koszinusz hasonlósági mérőszám alapján teljesen hasonlónak tekinthető, annak ellenére, hogy különböző értékelésekkel rendelkeznek.

````python
print('A értékelése')
print(a)
print('B értékelése')
print(b)
````

Ez egy gyakori jelenség a való világban, és az A felhasználóhoz hasonló felhasználókat kemény értékelőknek nevezhetjük.

Ilyen például egy filmkritikus, aki mindig az átlagnál alacsonyabb értékelést ad, de a listáján szereplő elemek rangsorolásában hasonló az olyan átlagos értékelőkéhez, mint a B.

### Az ajánlórendszerek típusai

Két fő típusa van:
- Kollaboratív szűrés:	Az emberek viselkedése alapján ajánl tartalmakat	 pl. Netflix, Amazon
- Tartalomalapú szűrés: Az ajánlandó elemek jellemzői alapján működik	pl. Spotify, YouTube
- Hybrid ajánlórendszerek: A fenti kettő kombinációja (Netflix, YouTube ezt használja).

### Kollaboratív szűrés (Collaborative Filtering)
- Az ajánlások a felhasználók viselkedése alapján készülnek.
- Feltételezés: Hasonló emberek hasonló dolgokat kedvelnek.
- Két altípusa:
 - Felhasználó-alapú (User-Based) -> Olyan felhasználókat keresünk, akik hasonlóan viselkednek.
 -  Tétel-alapú (Item-Based) -> Azok az elemek, amelyeket mások együtt fogyasztottak, valószínűleg együtt ajánlhatók.

Matematikai háttér:

- A hasonlóság mérése például koszinusz hasonlósággal vagy Pearson korrelációval történik.


Mi az a kollaboratív szűrés?
- Alapelv: "Ha két felhasználó korábban hasonlóan értékelt filmeket, akkor valószínűleg a jövőben is hasonló ízlésük lesz."
-  A felhasználói interakciók alapján ajánlunk tartalmakat, anélkül, hogy ismernénk a filmek tartalmát vagy jellemzőit.

Példa:


```
Film	      |  Felhasználó A	   | Felhasználó B	 | Felhasználó C
----------------------------------------------------------------
Inception	 |   5                  |     4	         |     5
Matrix	    |   4	              |     5             | 	?
Titanic       |   1                  | 	2             |	 1

```

- A és C hasonlóan értékelte az "Inception"-t és a "Titanic"-ot, így C valószínűleg 4 vagy 5-öt adna a "Matrix"-ra.


### Collaboratív szűrés távolságméréssel

Egyszerű megközelítés:
- Mérjük meg a hasonlóságot a felhasználók között koszinusz hasonlóság vagy Pearson korreláció segítségével.
- Megjósoljuk egy film értékelését a hasonló felhasználók értékelései alapján.

### User-Based kollaboratív szűrés (Koszinusz hasonlóság)
A hasonló felhasználók keresése koszinusz hasonlósággal.

````python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# 1. MovieLens dataset betöltése
url = 'https://raw.githubusercontent.com/rfarkas/student_data/main/MovieLens/movielens.small.ratings'
df = pd.read_csv(url, sep='\t', names=rating_colnames)
display(df.head())
````

````python
# 2. Pivot tábla készítése (Felhasználó x Film mátrix)
ratings_matrix = df.pivot(index='userID', columns='movieID', values='rating')
ratings_matrix
````

````python
# 3. Hasonlóság számítása a felhasználók között
user_similarity = cosine_similarity(ratings_matrix.fillna(0))

# 4. Eredmény megjelenítése Pandas DataFrame formában
user_sim_df = pd.DataFrame(user_similarity, index=ratings_matrix.index, columns=ratings_matrix.index)
display(user_sim_df.head(10))
````

Mit csinál ez a kód?

- Egy Felhasználó x Film mátrixot készítünk.
- Megmérjük a felhasználók közötti hasonlóságot koszinusz távolság alapján.
- A hasonló felhasználók segítenek megjósolni egy film értékelését.

````python
# User kiválasztása
user_id = 1
user_sim_df.loc[user_id,:].sort_values(ascending = False).head(5)
````

````python
import pandas as pd
from surprise import Dataset, Reader, SVD, KNNBasic
from surprise.model_selection import cross_validate, train_test_split
from surprise import accuracy

# 1. Adatok letöltése és előkészítése
# A Surprise saját Dataset.load_builtin() függvényével közvetlenül letölthetjük a MovieLens 100k datasetet.
data = Dataset.load_builtin('ml-100k')

# 2. Adathalmaz szétosztása tanító és teszt adatokra
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)
````

### Modell betanítása és kiértékelése
Használjunk egy KNNBasic alapú ajánlórendszert, amely egy mátrix-faktorizációs technika.

````python
# 3. KNNBasic modell létrehozása és betanítása
model = KNNBasic(k=3)
model.fit(trainset)

# 4. Előrejelzés a teszt adatokra
predictions = model.test(testset)

# 5. Kiértékelés – RMSE és MAE számítása
rmse = accuracy.rmse(predictions)
mae = accuracy.mae(predictions)
````

### Ajánlások generálása egy adott felhasználónak
Most ajánlásokat generálunk egy adott felhasználónak.

````python
#  6. Film ajánlás egy adott felhasználónak (pl. user ID = 196)
user_id = str(4)  # Surprise könyvtár stringként kezeli az ID-kat
all_items = [str(2)] #list(trainset.all_items())  # Összes film ID

#  7. Előrejelzések generálása minden filmre
predicted_ratings = [(item, model.predict(user_id, item).est) for item in all_items]

#  8. A 5 legjobb ajánlás kiválasztása
top_5_recommendations = sorted(predicted_ratings, key=lambda x: x[1], reverse=True)[:5]

#  9. Eredmények kiírása
print("Top 5 ajánlott film User 196 számára:")
for movie_id, predicted_rating in top_5_recommendations:
    print(f"Film ID: {movie_id}, Várható értékelés: {predicted_rating:.2f}")
````

### Tartalomalapú szűrés (Content-Based Filtering)
- Az ajánlás az ajánlott elemek jellemzői alapján történik.

- Kulcstechnológiák:
 - TF-IDF (Text Based ajánlás)
 - Feature Engineering (pl. filmek esetén műfaj, rendező, színészek)


```
Példa tartalomalapú szűrésre egy filmes adatbázison:
Film           |   Műfaj       |   Rendező |   Színész
----------------------------------------------------------
Inception      |   Sci-Fi      |   Nolan   |   DiCaprio
Interstellar   |   Sci-Fi      |   Nolan   |   McConaughey
Titanic        |   Romantikus  |   Cameron |   DiCaprio
```



Ha egy felhasználó szereti az "Inception"-t, akkor a rendszer ajánlhatja neki az "Interstellar"-t, mert hasonló műfajú és ugyanaz a rendező készítette.

### Tartalomalapú ajánlás TF-IDF segítségével

- A TF-IDF (Term Frequency - Inverse Document Frequency) egy szöveganalízis módszer, amely kiemeli a legfontosabb szavakat egy dokumentumból.
- A Cosine Similarity méri az egyes filmek közötti hasonlóságot a TF-IDF értékek alapján.

````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

#  1. Egyszerű filmadatbázis létrehozása
movies = pd.DataFrame({
    'title': ['Inception', 'Matrix', 'Interstellar', 'Titanic', 'Avatar', 'The Dark Knight'],
    'genre': ['Sci-Fi Action Thriller',
              'Sci-Fi Action',
              'Sci-Fi Drama',
              'Romantic Drama',
              'Sci-Fi Adventure',
              'Action Thriller']
})

#  2. TF-IDF vektorizáció a műfaj oszlopra
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(movies['genre'])
tfidf_matrix
````

````python
#  3. Koszinusz hasonlóság számítása az összes film között
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
cosine_sim
````

````python
#  4. Film ajánló függvény létrehozása
def recommend_movies(movie_title, cosine_sim=cosine_sim):
    idx = movies.index[movies['title'] == movie_title][0]  # Film indexe
    sim_scores = list(enumerate(cosine_sim[idx]))  # Hasonlósági értékek
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)  # Legjobban hasonlók kiválasztása
    sim_scores = sim_scores[1:4]  # Top 3 hasonló film

    # 5. Filmcímek lekérése
    movie_indices = [i[0] for i in sim_scores]
    return movies.iloc[movie_indices][['title', 'genre']]
````

````python
# 6. Ajánlások kérése egy adott filmhez
recommended_movies = recommend_movies('Inception')
display(recommended_movies)
````

Hogyan működik ez?
- A TF-IDF vektorizálja a film műfajokat, és egy numerikus reprezentációt hoz létre.
- A Cosine Similarity méri, hogy melyik film mennyire hasonlít a másikra.
- A legjobban hasonlító filmeket ajánljuk az adott filmhez.

### Hibrid ajánlórendszer magyarázata és példája
- Megérteni, hogyan működik egy hibrid ajánlórendszer (Hybrid Recommender System), amely ötvözi a kollaboratív és tartalomalapú szűrést.
- Surprise könyvtár (SVD modell) + TF-IDF (tartalomalapú szűrés).



### Mi az a hibrid ajánlórendszer?
- Kombinálja a kollaboratív szűrést és a tartalomalapú szűrést, így pontosabb ajánlásokat ad.
- Kollaboratív szűrés – Más felhasználók értékelései alapján ajánlunk.
- Tartalomalapú szűrés – A filmek jellemzői alapján ajánlunk.
- Hibrid modell: Mindkettőt kombinálja, így ha nincs sok felhasználói adat, még mindig képes ajánlani tartalom alapján.

Példa:
```
Film        |  Műfajok          |  Felhasználó A értékelése |  Felhasználó B értékelése
--------------------------------------------------------------------------------
Inception   |  Sci-Fi, Akció    |  5                        |  4
Matrix      |  Sci-Fi, Akció    |  4                        |  5
Interstellar|  Sci-Fi, Dráma    |  ?                        |  5
Titanic     |  Romantikus, Dráma|  1                        |  2
```

- Ha A felhasználó szerette az Inception-t, akkor ajánlhatunk neki a Matrix-ot kollaboratív szűréssel.
- Ha nincs elég értékelési adat, akkor ajánlhatunk az "Interstellar"-t, mert tartalmilag hasonló az Inception-höz.



````python
import pandas as pd

rating_colnames = ['userID', # felhasználó azonosítója
                   'movieID', # film azonosítója
                   'unix_timestamp', # időpont amikor a felhasználó értékelte a filmet
                   'rating' # értékelés 1-5 skálán, 5 a legjobb
                   ]

# 1. MovieLens dataset betöltése
url = 'https://raw.githubusercontent.com/rfarkas/student_data/main/MovieLens/movielens.small.ratings'
df = pd.read_csv(url, sep='\t', names=rating_colnames)
display(df.head())
````

````python
import pandas as pd
import numpy as np
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# 2. Kollaboratív szűrés (SVD modell a felhasználói értékelésekkel)
reader = Reader(rating_scale=(1, 5))
data = Dataset.load_from_df(df[['userID', 'movieID', 'rating']], reader)
data
````

````python
# 3. Tanító és tesztadatok elkülönítése
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)

# 4. SVD modell betanítása
svd = SVD()
svd.fit(trainset)
````

````python
# 5. Filmadatok betöltése és TF-IDF modell készítése a műfaj alapján
movies = pd.DataFrame({
    'movie_id': [1, 2, 3, 4, 5, 6],  # Példa ID-k
    'title': ['Inception', 'Matrix', 'Interstellar', 'Titanic', 'Avatar', 'The Dark Knight'],
    'genre': ['Sci-Fi Action Thriller',
              'Sci-Fi Action',
              'Sci-Fi Drama',
              'Romantic Drama',
              'Sci-Fi Adventure',
              'Action Thriller']
})

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(movies['genre'])
````

````python
# 6. Hibrid ajánló függvény
def hybrid_recommend(user_id, movie_title):
    # 6.1. Kollaboratív szűrés – Felhasználó-film értékelés becslése
    movie_id = movies[movies['title'] == movie_title]['movie_id'].values[0]
    collaborative_score = svd.predict(user_id, movie_id).est

    # 6.2. Tartalomalapú szűrés – Hasonló filmek keresése
    idx = movies.index[movies['title'] == movie_title][0]  # Film indexe
    sim_scores = list(enumerate(cosine_similarity(tfidf_matrix[idx], tfidf_matrix)[0]))  # Hasonlóságok
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)  # Legjobban hasonlók
    sim_scores = sim_scores[1:4]  # Top 3 hasonló film

    # 6.3. Filmcímek lekérése
    movie_indices = [i[0] for i in sim_scores]
    content_based_recommendations = movies.iloc[movie_indices][['title', 'genre']]

    return collaborative_score, content_based_recommendations
````

````python
# 7. Ajánlások egy adott felhasználónak egy film alapján
user_id = 100  # Példa felhasználó
movie_title = "Inception"

collab_score, content_rec = hybrid_recommend(user_id, movie_title)

# 8. Eredmények kiírása
print(f"A felhasználó ({user_id}) várható értékelése '{movie_title}'-re: {collab_score:.2f}")
print("\nHasonló filmek a tartalom alapján:")
display(content_rec)
````

### Hogyan működik ez?
- Kollaboratív szűrés (SVD) segítségével előrejelez egy értékelést egy felhasználónak egy filmre.
- Ha nincs elég adat, a tartalomalapú szűrés (TF-IDF + Cosine Similarity) segít ajánlani.
- A két megoldást kombináljuk, így ha nincs sok értékelési adat, akkor is tudunk ajánlani filmeket.
