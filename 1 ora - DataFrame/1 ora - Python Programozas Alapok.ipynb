## Változók és típusok

A Pythonban nem kell előre megadni a változó típusát, a hozzárendelt érték alapján jön létre.
Alapvető típusok: `int`, `float`, `str`, `list`, `tuple`, `dict`, `set`, `bool`, `None`.

```python
a = 42
b = 'asd'
c = "asd"
d = f"{a} is a number"  # f-string
d2 = "%d is a number" % a  # régi formázás
e, f = 5, 6  # több változó egyszerre
f, e = e, f  # csere
print(a, b, c, d, d2, e, f)
```

## Műveletek

Pythonban elérhetők a szokásos aritmetikai, logikai és bitműveletek.
- `**` hatványozás
- `//` egész osztás
- `&`, `|`, `^`, `<<`, `>>` bitműveletek
- `math` modul speciális függvényei: `ceil`, `floor`, `round`

```python
import math

print("3 ** 2 =", 3 ** 2)
print("3 // 2 =", 3 // 2)
print("ceil(3.8):", math.ceil(3.8))
print("floor(3.8):", math.floor(3.8))
print("round(3.8):", round(3.8))
print("round(3.8,1):", round(3.8, 1))

a = 10; b = 5
print("a & b:", a & b)
print("a | b:", a | b)
print("a ^ b:", a ^ b)
print("a << 1:", a << 1)
print("a >> 1:", a >> 1)
print("~a:", ~a)
```

## String műveletek

A sztringek `str` típusúak, rengeteg beépített metódussal rendelkeznek.
Tipikus műveletek: nagybetűsítés, kicsinyítés, whitespace levágása, cserék, darabolás.

```python
b = "   I'm a String   "
print("upper:", b.upper())
print("lower:", b.lower())
print("strip:", b.strip())
print("replace:", b.replace("a", "b"))
print("split('s'):", b.split("s"))
```

## Listák

A lista (`list`) többféle elemet tartalmazhat, dinamikusan bővíthető, módosítható.
Fontos metódusok: `append`, `insert`, `extend`, `remove`, `del`, szeletelés (slicing).

```python
l = [1, 2, 'a', [5, 6]]
print("l[3]:", l[3])
print("l[-1]:", l[-1])
l.append("b")
l.insert(2, "b")
l.extend([1, 2, 3])
print("l contains 3?", 3 in l)
l.remove("b")
del l[2]
print("After removals:", l)

# slicing példák
print("l[1:]:", l[1:])
print("l[:-1]:", l[:-1])
print("l[1:-1:2]:", l[1:-1:2])
```

## Tuple

A tuple hasonló a listához, de **nem módosítható** (immutable).
Használható például fix adatok tárolására.

```python
t = (1, 'a', "c", 3)
print("Tuple:", t)
# t[2] = 'b'  # hibát dobna, mert a tuple nem módosítható
```

## Dictionary (szótár)

Kulcs-érték párokat tárol, gyors keresést biztosít.
Fontos metódusok: `keys()`, `values()`, `items()`.

```python
team = {
    91: "Ayers, Robert",
    13: "Beckham Jr,",
    3: "Brown, Josh",
    54: "Casillas, Jonathan",
    21: "Collins, Landon"
}
print("len:", len(team))
print("team[3]:", team[3])
team[2] = "Brown"
del team[2]
print("keys:", list(team.keys()))
print("values:", list(team.values()))
print("items:", list(team.items()))
```

## Generátor jellegű kifejezések

Pythonban röviden létrehozhatunk listát, szótárat vagy tuple-t generátorral (`comprehension`).

```python
mylist = [x * x for x in range(10)]
mydict = {x: x * x for x in range(5)}
mydict2 = {x: x * x for x in range(5) if x != 2}
mytuple = tuple(x * x for x in range(3))
print(mylist, mydict, mydict2, mytuple)
```

## Elágazások és ciklusok

Pythonban `if/elif/else` szerkezetet használunk döntésekhez.
Ismétlésre `for` és `while` ciklus áll rendelkezésre.

```python
if 100 in team:
    print('Yes, 100 is in the team')
elif 76 in team:
    print('100 is not in the team, but 76 is...')
else:
    print('Both 100 and 76 are not in the team')

for num, name in team.items():
    print(num, name, end="\t")
print()

for i in range(2, 10, 2):
    print(i)
else:
    print("Last i:", i)

i = 1
while i < 10:
    print(i)
    i += 1
else:
    print("i is now:", i)
```

## Függvények

A függvények `def` kulcsszóval definiálhatók.
Lehet alapértelmezett paraméterük, és visszatérhetnek értékkel.

```python
def is_even(num: int = 5):
    return num % 2 == 0

def is_even2(num: int = 5):
    return (num % 2 == 0, num)

print(is_even2())
```

## Globális változók

A `global` kulcsszóval egy függvényen belül a külső változó is módosítható.
Alapértelmezésben a függvényen belül létrehozott azonosító **lokális** változó.

```python
x = "awesome"

def myfunc():
    x = "fantastic"  # lokális
    print("Python is " + x)

def myfunc_global():
    global x
    x = "fantastic"  # módosítja a globális változót
    print("Inside:", x)

myfunc()
print("Outside after myfunc():", x)
myfunc_global()
print("Outside after myfunc_global():", x)
```
