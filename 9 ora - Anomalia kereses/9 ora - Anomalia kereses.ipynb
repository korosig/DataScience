# Anomáliák Detektálása

## Mi az anomáliák detektálása?
Az anomáliák detektálása (más néven outlier detection) az adattudomány egyik fontos területe, amely célja az olyan adatpontok vagy események azonosítása, amelyek jelentősen eltérnek a normális viselkedéstől. Az anomáliák azonosítása kulcsfontosságú lehet például:
- Pénzügyi csalások detektálása
- Rendszerbiztonsági problémák felismerése
- Egészségügyi monitorozás és diagnosztika
- Minőségellenőrzés

## Anomália vs Outlier
Bár gyakran szinonimaként használják, az anomáliák és outlierek közötti különbség abban rejlik, hogy az anomáliák gyakran az egész rendszer működését befolyásoló szokatlan eseményeket jelzik, míg az outlierek egyszerűen csak olyan adatpontok, amelyek eltérnek az átlagtól, de nem biztos, hogy kritikusak.

## Módszerek az anomáliák detektálására

### 1. **Statisztikai alapú módszerek**
A statisztikai alapú módszerek gyakran a **Z-score** vagy az **Interkvartilis tartomány (IQR)** alapú technikákra építenek.
- **Z-score**: Az adatpontok eltérése az átlagtól, szórásos mértékben. Ha a Z-score nagyobb, mint egy előre meghatározott küszöb (pl. 3), akkor az adatot anomáliaként kezelhetjük.
- **IQR (Interquartile Range)**: A minta középső 50%-át lefedő tartomány. Ha egy adatpont túlmutat ezen a tartományon, akkor anomáliának számíthat.

### 2. **Gépi tanulás alapú módszerek**
A gépi tanulás alapú módszerek képesek komplexebb anomáliákat felismerni, például:
- **Klaszterezés alapú módszerek (K-means, DBSCAN)**: A klaszterekből való kilógás jelzi az anomáliákat.
- **Autoencoderek**: A neurális hálózati autoencoderek az adatok tömörítésére és rekonstruálására használhatók. Az anomáliák azonosíthatók, ha az autoencoder nem képes jól rekonstruálni az adatpontot.
- **Isolation Forest**: Egy döntési fákon alapuló módszer, amely hatékony az anomáliák észlelésében, különösen nagy adathalmazok esetén.

### 3. **Idősoros adatok**
Idősorokban az anomáliák gyakran hirtelen, szokatlan változásokat jelentenek.
- **ARIMA modellek**: Az időbeli adatok előrejelzésére és az anomáliák detektálására használható.
- **LSTM (Long Short-Term Memory)**: A mély tanulás alapú módszerek közül az LSTM különösen alkalmas az idősoros adatok kezelésére és az anomáliák felismerésére.

### 4. **Egyéb módszerek**
- **One-Class SVM**: Különösen hasznos, ha csak a normális adatokat ismerjük, és az anomáliák észlelésére van szükség.

## Alkalmazások
Az anomáliák detektálása széles körben alkalmazható:
- **Pénzügyi szektor**: Csalások, szokatlan tranzakciók azonosítása.
- **Egészségügy**: Hirtelen változások a betegek életjeleiben (pl. szívritmus).
- **Ipar**: Gépek, eszközök működésének figyelése és meghibásodások előrejelzése.
- **Hálózati biztonság**: Szokatlan hálózati tevékenység figyelése.



# Anomáliák Detektálása: IQR és Z-score

## Mi az anomáliák detektálása?
Az anomáliák detektálása (vagy outlier detection) az adattudomány egyik fontos eszköze, amely lehetővé teszi az adathalmazban található szokatlan vagy eltérő adatpontok azonosítását. Az anomáliák gyakran fontos információkat hordoznak, mint például a pénzügyi csalások, rendszerhibák vagy orvosi anomáliák.

## IQR alapú anomáliák keresése
Az **Interkvartilis tartomány (IQR)** alapú módszer az adatok eloszlását vizsgálja. Az alsó (Q1) és felső (Q3) kvartilisek közötti különbséget, az IQR-t használjuk annak meghatározására, hogy mely adatok esnek kívül ezen a tartományon, és azokat tekinthetjük anomáliának.

### IQR lépések:
1. Kiszámítjuk az első (Q1) és harmadik (Q3) kvartilist.
2. Meghatározzuk az IQR-t: `IQR = Q3 - Q1`.
3. Az anomáliákra a következő küszöbértékeket használjuk:
   - **Alsó küszöb**: `Q1 - 1.5 * IQR`
   - **Felső küszöb**: `Q3 + 1.5 * IQR`

## Z-score alapú anomáliák keresése
A **Z-score** egy statisztikai mérőszám, amely megmondja, hogy egy adatpont hány szórásnyira van az átlagtól. Azokat az adatokat tekintjük anomáliának, amelyeknek a Z-score-ja meghaladja a 3-at (vagy más előre meghatározott küszöböt).

### Z-score lépések:
1. Kiszámítjuk az átlagot és szórást az adathalmazon.
2. A Z-score-t minden adatpontra kiszámítjuk: `Z = (adatpont - átlag) / szórás`.
3. Azokat az adatpontokat tekintjük anomáliának, amelyek Z-score-ja meghaladja a 3-at.

## Alkalmazás: Magasság, Testsúly és Életkor
Ebben az elemzésben három jellemzőn (magasság, testsúly, életkor) alkalmazzuk az IQR és Z-score alapú anomáliák keresését. Az adathalmazon szándékosan hozzáadott extrém értékek (anomáliák) segítenek bemutatni a két módszer működését.

## Vizualizáció
A módszerek alkalmazása után vizualizáljuk az adatokat, hogy könnyen észrevegyük az anomáliákat mindkét módszer alapján. Az eredményeket ábrázoljuk, hogy jobban megértsük, hogyan működnek az anomáliák detektálásának technikái.

## Következő lépések
Próbáljuk meg az IQR és Z-score módszereket különböző adatkészleteken is, hogy felmérjük, melyik módszer működik jobban különböző típusú anomáliák esetén.


```python
### Példa: Z-score alapú anomáliák keresése
import numpy as np
import pandas as pd

# Példa adat
data = [10, 12, 13, 15, 100, 16, 18, 19, 20, 21]
```

```python
# az adathalmaz ábrázolása
pd.DataFrame(data).plot()
plt.title('A nyers jelünk')
```

```python
# Z-score számítása
mean = np.mean(data)
std_dev = np.std(data)
z_scores = [(x - mean) / std_dev for x in data]

# Anomáliák detektálása (pl. Z-score > 2 vagy < -2)
anomalies = [data[i] for i, z in enumerate(z_scores) if abs(z) > 2]

print(f"Anomáliák: {anomalies}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generáljunk egy egyszerű adat halmazt
np.random.seed(42)
data = np.random.normal(0, 1, 100)  # Normál eloszlású adatok
data_with_outliers = np.append(data, [10, 12, -8, -10])  # Kivételek hozzáadása
```

```python
# az adathalmaz ábrázolása
pd.DataFrame(data_with_outliers).plot()
plt.title('A nyers jelünk')
```

```python
# IQR kiszámítása
Q1 = np.percentile(data_with_outliers, 25)  # 25. percentilis (1. kvartilis)
Q3 = np.percentile(data_with_outliers, 75)  # 75. percentilis (3. kvartilis)
IQR = Q3 - Q1  # Interkvartilis tartomány

# Anomália küszöbök
lower_bound = Q1 - 1.5 * IQR  # Alsó küszöb
upper_bound = Q3 + 1.5 * IQR  # Felső küszöb

# Anomáliák detektálása
anomalies = [x for x in data_with_outliers if x < lower_bound or x > upper_bound]

# Vizualizáció
plt.figure(figsize=(10, 6))
plt.boxplot(data_with_outliers, vert=False, patch_artist=True, boxprops=dict(facecolor='skyblue', color='blue'))
plt.scatter(anomalies, np.ones(len(anomalies)), color='red', label='Anomáliák', zorder=5)
plt.title('IQR alapú anomáliák detektálása')
plt.xlabel('Érték')
plt.legend()
plt.show()

print(f"Anomáliák: {anomalies}")
```

```python
# az adathalmaz ábrázolása anoláliák nélkül
data_without_outliers = [x for x in data_with_outliers if x >= lower_bound and x <= upper_bound]
pd.DataFrame(data_without_outliers).plot()
plt.title('A jelünk anomália nélkül')
```

## Anomáliák detektálása a többdienziós térben

Ebben a példában két klasszikus statisztikai módszert használunk az **anomáliák (outlierek)** felismerésére egy egyszerű, mesterségesen generált adathalmazon.  
A dataset három emberi paramétert tartalmaz: **magasság**, **testsúly** és **életkor**, valamint néhány szándékosan beillesztett extrém értéket.


## Mit fogunk látni?
- Egy **szintetikusan generált** adathalmazt (100 normál és néhány extrém érték).  
- Két különböző módszert az anomáliák detektálására:
  - **IQR-alapú** detektálás
  - **Z-score-alapú** detektálás
- Több scatter ploton ábrázoljuk:
  - A normális adatokat (kék / zöld / narancs pontok)
  - Az anomáliákat (piros pontok)
- A végén a program kiírja az azonosított anomáliákat.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Generáljunk egy egyszerű adatot 3 paraméterrel: magasság (cm), testsúly (kg), életkor (év)
np.random.seed(42)
heights = np.random.normal(170, 10, 100)  # Magasság (170 cm átlag, 10 cm szórás)
weights = np.random.normal(70, 15, 100)   # Testsúly (70 kg átlag, 15 kg szórás)
ages = np.random.normal(30, 5, 100)       # Életkor (30 év átlag, 5 év szórás)

```

```python
# Adatokba szándékosan beszúrunk néhány anomáliát
heights = np.append(heights, [250, 55])  # Extrém magasságok
weights = np.append(weights, [200, 30])  # Extrém testsúlyok
ages = np.append(ages, [5, 85])          # Extrém életkorok

# Hozzuk létre a DataFrame-et
data = pd.DataFrame({'Height': heights, 'Weight': weights, 'Age': ages})
```

```python
# IQR alapú anomáliák keresése
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# IQR alapú anomáliák
IQR_anomalies = data[(data < lower_bound) | (data > upper_bound)].dropna()
```

```python
# Z-score alapú anomáliák keresése
z_scores = np.abs(zscore(data))
z_threshold = 3  # Z-score > 3 esetén anomáliát keresünk

# Z-score alapú anomáliák
zscore_anomalies = data[(z_scores > z_threshold)]
```

```python
# Plotoljuk az adatokat, és kiemeljük az anomáliákat
plt.figure(figsize=(12, 8))

# Magasság plotolása
plt.subplot(3, 1, 1)
plt.scatter(range(len(data['Height'])), data['Height'], label='Magasság', color='blue')
plt.scatter(IQR_anomalies.index, IQR_anomalies['Height'], color='red', label='IQR Anomália')
plt.title('Magasság (cm)')
plt.legend()
# Testsúly plotolása
plt.subplot(3, 1, 2)
plt.scatter(range(len(data['Weight'])), data['Weight'], label='Testsúly', color='green')
plt.scatter(IQR_anomalies.index, IQR_anomalies['Weight'], color='red', label='IQR Anomália')
plt.title('Testsúly (kg)')
plt.legend()
# Életkor plotolása
plt.subplot(3, 1, 3)
plt.scatter(range(len(data['Age'])), data['Age'], label='Életkor', color='orange')
plt.scatter(IQR_anomalies.index, IQR_anomalies['Age'], color='red', label='IQR Anomália')
plt.title('Életkor (év)')
plt.legend()
plt.tight_layout()
plt.show()

```

```python
# Z-score anomáliák plotolása
plt.figure(figsize=(12, 8))
# Magasság plotolása
plt.subplot(3, 1, 1)
plt.scatter(range(len(data['Height'])), data['Height'], label='Magasság', color='blue')
plt.scatter(zscore_anomalies.index, zscore_anomalies['Height'], color='red', label='Z-score Anomália')
plt.title('Magasság (cm) - Z-score Anomáliák')
plt.legend()
# Testsúly plotolása
plt.subplot(3, 1, 2)
plt.scatter(range(len(data['Weight'])), data['Weight'], label='Testsúly', color='green')
plt.scatter(zscore_anomalies.index, zscore_anomalies['Weight'], color='red', label='Z-score Anomália')
plt.title('Testsúly (kg) - Z-score Anomáliák')
plt.legend()
# Életkor plotolása
plt.subplot(3, 1, 3)
plt.scatter(range(len(data['Age'])), data['Age'], label='Életkor', color='orange')
plt.scatter(zscore_anomalies.index, zscore_anomalies['Age'], color='red', label='Z-score Anomália')
plt.title('Életkor (év) - Z-score Anomáliák')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Kiíratjuk az anomáliákat
print(f"IQR alapú anomáliák:\n{IQR_anomalies}")
print(f"\nZ-score alapú anomáliák:\n{zscore_anomalies}")
```

# Klaszterezés Alapú Anomáliák Detektálása

## Mi a klaszterezés alapú anomáliák detektálása?
A klaszterezés alapú anomáliák detektálása olyan technika, amelyet akkor alkalmazunk, amikor az adatpontok természetes csoportokba sorolhatók. Az **anomáliákat** azok az adatpontok jelentik, amelyek nem illeszkednek jól egyetlen klaszterhez sem, vagyis nagy távolságban vannak a klaszter középpontjától.

## Hogyan működik a K-means klaszterezés?
A **K-means** egy népszerű klaszterezési algoritmus, amely az adatokat K darab klaszterbe sorolja. A klaszterek középpontjait (centroidjait) használja az adatpontok csoportosítására. Az algoritmus célja, hogy minimalizálja az adatpontok közötti távolságot a klaszter középpontokhoz.

## Anomáliák Detektálása
A klaszterezés alapú anomáliák detektálásában azokat az adatokat tekintjük anomáliának, amelyek túl nagy távolságra helyezkednek el a klaszter középpontjától. Az ilyen adatpontok a legtöbb klaszterhez nem illeszkednek jól, és szokatlan mintázatokat képviselnek.


## Alkalmazás: Magasság, Testsúly és Életkor
Ebben a példában három jellemzőt (magasság, testsúly, életkor) használtunk. A K-means algoritmus segítségével klaszterekbe soroltuk az adatokat, és a távolság alapján detektáltuk az anomáliákat. A piros pontok jelölik azokat az adatokat, amelyek az anomáliákat képviselik.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Generáljunk egy egyszerű adatot 3 paraméterrel: magasság (cm), testsúly (kg), életkor (év)
np.random.seed(42)
heights = np.random.normal(170, 10, 100)  # Magasság (170 cm átlag, 10 cm szórás)
weights = np.random.normal(70, 15, 100)   # Testsúly (70 kg átlag, 15 kg szórás)
ages = np.random.normal(30, 5, 100)       # Életkor (30 év átlag, 5 év szórás)
```

```python
# Adatokba szándékosan beszúrunk néhány anomáliát
heights = np.append(heights, [250, 55])  # Extrém magasságok
weights = np.append(weights, [200, 30])  # Extrém testsúlyok
ages = np.append(ages, [5, 85])          # Extrém életkorok

# Hozzuk létre a DataFrame-et
data = pd.DataFrame({'Height': heights, 'Weight': weights, 'Age': ages})
```

```python
# Skálázás (fontos, hogy a K-means algoritmus érzékeny a különböző skálájú változókra)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
```

```python
# K-means klaszterezés (3 klasztert választunk)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(scaled_data)
```

```python

# Klaszterek és a legközelebbi középpontok kiszámítása
data['Cluster'] = kmeans.labels_
data['Distance_from_centroid'] = [np.linalg.norm(x - kmeans.cluster_centers_[label]) for x, label in zip(scaled_data, kmeans.labels_)]
```

```python
# Az anomáliák a legnagyobb távolságú pontok lesznek (pl. az eltérő klaszterekhez tartozó pontok)
threshold = np.percentile(data['Distance_from_centroid'], 95)  # A legnagyobb 5%-ot tekintjük anomáliának
anomalies = data[data['Distance_from_centroid'] > threshold]
```

```python
# Vizualizálás
plt.figure(figsize=(10, 6))
plt.scatter(data['Height'], data['Weight'], c=data['Cluster'], cmap='viridis', label='Adatok')
plt.scatter(anomalies['Height'], anomalies['Weight'], color='red', label='Anomáliák', s=100)
plt.title('Klaszterezés alapú anomáliák detektálása (K-means)')
plt.xlabel('Magasság (cm)')
plt.ylabel('Testsúly (kg)')
plt.legend()
plt.show()

# Kiíratjuk az anomáliákat
print(f"Anomáliák:\n{anomalies[['Height', 'Weight', 'Age', 'Distance_from_centroid']]}")
```

### Megjegyzés az ábrához

A **K-Means** algoritmus a pontokat **többdimenziós térben** (jelen esetben: magasság, testsúly és életkor) osztja klaszterekbe úgy, hogy minimalizálja az egyes pontok és a klaszterközéppontok közötti távolságot.  

A fenti ábrán azonban csak a **magasság–testsúly síkot** ábrázoljuk, így a vizualizáció **nem tükrözi teljesen** a modell által figyelembe vett összes dimenziót.  
Lehetséges tehát, hogy egy adatpont a 2D síkon „másik csoportba” tartozónak tűnik, de a **3D térben (ahol az életkor is szerepel)** már a modell által helyesen besorolt klaszterbe esik.

Összefoglalva:
- A K-Means a **teljes dimenziós térben** számít távolságokat (pl. 3D-ben: magasság–testsúly–életkor).  
- A 2D ábrázolás csak **egy vetületet** mutat a valós döntési térből.  
- Ezért előfordulhat, hogy egy pont vizuálisan nem illik a csoportjába, de matematikailag (a harmadik dimenzióval együtt) **jó helyen van**.




# Isolation Forest - Anomáliák Detektálása

## Mi az Isolation Forest?
Az **Isolation Forest** egy hatékony algoritmus az anomáliák detektálására, amelyet különösen nagy adathalmazok esetén használunk. Az algoritmus célja, hogy az adatokat döntési fák segítségével "elszigetelje", és azokat az adatpontokat tekintjük anomáliának, amelyek gyorsan elszigetelhetők, mivel eltérnek a többi adatponttól.

## Hogyan működik az Isolation Forest?
Az **Isolation Forest** a következőképpen működik:
1. **Döntési fák** segítségével az adatpontokat elválasztja a többi adatponttól.
2. Azokat az adatpontokat, amelyek gyorsan elszigetelhetők, anomáliaként jelöli ki.
3. Az anomáliák nem illeszkednek jól a normális adatpontok közé, így azok nagyobb valószínűséggel kerülnek elszigetelésre.



## Alkalmazás: Magasság, Testsúly és Életkor
Ebben a példában három jellemzőt (magasság, testsúly, életkor) használtunk, és az **Isolation Forest** segítségével detektáltuk az anomáliákat. Az anomáliák piros pontokkal vannak kiemelve a vizualizáción.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Generáljunk egy egyszerű adatot 3 paraméterrel: magasság (cm), testsúly (kg), életkor (év)
np.random.seed(42)
heights = np.random.normal(170, 10, 100)  # Magasság (170 cm átlag, 10 cm szórás)
weights = np.random.normal(70, 15, 100)   # Testsúly (70 kg átlag, 15 kg szórás)
ages = np.random.normal(30, 5, 100)       # Életkor (30 év átlag, 5 év szórás)

# Adatokba szándékosan beszúrunk néhány anomáliát
heights = np.append(heights, [250, 55])  # Extrém magasságok
weights = np.append(weights, [200, 30])  # Extrém testsúlyok
ages = np.append(ages, [5, 85])          # Extrém életkorok

# Hozzuk létre a DataFrame-et
data = pd.DataFrame({'Height': heights, 'Weight': weights, 'Age': ages})

```

```python
# Skálázás (fontos, hogy a modellen érzékeny a változók skálájára)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
```

```python
# Isolation Forest alkalmazása
model = IsolationForest(contamination=0.1, random_state=42)  # contamination: várható anomáliák aránya
model.fit(scaled_data)
```

```python
# Anomáliák detektálása
predictions = model.predict(scaled_data)
# Az anomáliák -1-ként jelennek meg, a normális adatpontok 1-ként
data['Anomaly'] = predictions
```

```python
# Vizualizáció
plt.figure(figsize=(10, 6))

# Magasság és Testsúly vizualizálása
plt.scatter(data['Height'], data['Weight'], c=data['Anomaly'], cmap='coolwarm', label='Adatok')
plt.title('Isolation Forest - Anomáliák Detektálása')
plt.xlabel('Magasság (cm)')
plt.ylabel('Testsúly (kg)')
plt.legend()
plt.show()

# Kiíratjuk az anomáliákat
anomalies = data[data['Anomaly'] == -1]
print(f"Anomáliák:\n{anomalies[['Height', 'Weight', 'Age']]}")
```

### Megjegyzés az ábrához

Az **Isolation Forest** algoritmus valójában **többdimenziós térben** (jelen esetben: magasság, testsúly és életkor) tanulja meg az adatok eloszlását, és ott próbálja **elszigetelni az anomáliákat**.  

A fenti ábra viszont csak a **magasság–testsúly síkon** mutatja az adatokat, ezért a vizualizáció **nem teljesen fedi le** a modell valódi döntési terét.  
Lehetséges, hogy egy adatpont a 2D ábrán "normálisnak" tűnik, de a **3D térben** (ahol az életkort is figyelembe veszi a modell) **anomáliaként** azonosítódik — és fordítva is előfordulhat.

Összefoglalva:
- Az Isolation Forest **multidimenziós döntési fák** alapján dolgozik.  
- A 2D ábrázolás csak **részleges vetületet** mutat a modell működéséből.  
- Az anomáliák tehát nemcsak az érték nagysága, hanem **az összefüggések mintázata** alapján is detektálódnak.


# One-Class SVM - Anomáliák Detektálása

## Mi a One-Class SVM?
A **One-Class SVM** (Support Vector Machine) egy olyan változata a hagyományos SVM-nek, amelyet az anomáliák detektálására használnak, amikor csak a normális adataink állnak rendelkezésre. A módszer a normális adatokat használja arra, hogy egy döntési határt (hipersíkot) találjon, amely legjobban "elhatárolja" a normális adatokat. Azokat az adatokat tekintjük anomáliáknak, amelyek kívül esnek ezen a határon.

## Hogyan működik a One-Class SVM?
1. **Tanulás**: A One-Class SVM csak a normális adatokat használja fel a döntési határ meghatározására.
2. **Predikció**: Az új adatokat a modell a megtanult döntési határ alapján normálisnak vagy anomáliának minősíti.
3. **Anomáliák**: Azokat az adatokat tekintjük anomáliáknak, amelyek a döntési határon kívül helyezkednek el.


## Alkalmazás: Magasság, Testsúly és Életkor
Ebben az alkalmazásban három jellemzőt (magasság, testsúly, életkor) használtunk. Az **One-Class SVM** segítségével detektáltuk az anomáliákat, és azokat a pontokat, amelyek eltérnek a normális eloszlástól, anomáliaként jelöltük.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler

# Generáljunk egy egyszerű adatot 3 jellemzővel: magasság (cm), testsúly (kg) és életkor (év)
np.random.seed(42)
heights = np.random.normal(170, 10, 100)  # Magasság (átlag 170 cm, szórás 10 cm)
weights = np.random.normal(70, 15, 100)   # Testsúly (átlag 70 kg, szórás 15 kg)
ages = np.random.normal(30, 5, 100)       # Életkor (átlag 30 év, szórás 5 év)

# Szándékosan hozzáadunk néhány extrém értéket (anomáliát)
heights = np.append(heights, [250, 55])  # Extrém magasságok
weights = np.append(weights, [200, 30])  # Extrém testsúlyok
ages = np.append(ages, [5, 85])          # Extrém életkorok

# Hozzuk létre a DataFrame-et
data = pd.DataFrame({'Height': heights, 'Weight': weights, 'Age': ages})
```

```python
# Az adatok skálázása (fontos, mert a One-Class SVM érzékeny a jellemzők méretarányára)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
```

```python
# One-Class SVM modell tanítása
model = OneClassSVM(nu=0.1, kernel="rbf", gamma='scale')  # nu: a várható anomáliák aránya
model.fit(scaled_data)
```

```python
# Anomáliák előrejelzése (-1 = anomália, 1 = normális adat)
predictions = model.predict(scaled_data)
data['Anomaly'] = predictions
```

```python
# Eredmények vizualizálása
plt.figure(figsize=(10, 6))
# Magasság és testsúly megjelenítése, az anomáliák kiemelésével
plt.scatter(data['Height'], data['Weight'], c=data['Anomaly'], cmap='coolwarm', label='Adatok')
plt.title('One-Class SVM - Anomáliák Detektálása')
plt.xlabel('Magasság (cm)')
plt.ylabel('Testsúly (kg)')
plt.legend()
plt.show()

# Anomáliák kiíratása
anomalies = data[data['Anomaly'] == -1]
print(f"Detektált anomáliák:\n{anomalies[['Height', 'Weight', 'Age']]}")
```

## Megjegyzés a vizualizációhoz
A One-Class SVM valójában **háromdimenziós térben** (magasság–testsúly–életkor) tanulja a döntési határt.  
A fenti ábrán azonban csak a **magasság–testsúly sík** vetületét látjuk, így:
- az algoritmus **többdimenziós döntési határát** nem lehet pontosan megjeleníteni,  
- lehetséges, hogy egy pont 2D-ben normálisnak tűnik, de 3D-ben anomália (vagy fordítva).

# Idősoros Anomáliák Detektálása

## Mi az idősoros anomáliák detektálása?
Az **idősoros anomáliák detektálása** célja olyan adatpontok azonosítása, amelyek szokatlanok, és eltérnek a normál mintázattól. Az anomáliák lehetnek például váratlan csúcsok, mélypontok, vagy bármilyen hirtelen változás az adatokban. Ez különösen fontos a pénzügyi előrejelzésekben, az ipari rendszerekben és az egészségügyi monitorozásban.

## Módszerek az idősorok anomáliáinak keresésére
Két népszerű módszert használunk az anomáliák keresésére:
1. **Isolation Forest**: Az **Isolation Forest** az adatokat döntési fák segítségével "elszigeteli", és azokat az adatokat tekinti anomáliának, amelyek nem illeszkednek a normális mintázathoz.
2. **Z-score**: A **Z-score** segítségével meghatározzuk, hogy egy adatpont hány szórásnyira van az átlagtól. Ha az érték nagyobb, mint egy küszöb (pl. 3), akkor azt anomáliaként tekintjük.



## Alkalmazás: Hőmérséklet Idősor
Ebben a példában egy szintetikus idősor adatot használunk, amely napi hőmérsékleteket tartalmaz. Az **Isolation Forest** és **Z-score** módszerek segítségével detektáljuk a szokatlan hőmérsékleti adatokat.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from scipy.stats import zscore

# Generáljunk egy szintetikus idősor adatot (pl. napi hőmérséklet)
np.random.seed(42)
days = 365  # 1 év
temperatures = np.random.normal(20, 5, days)  # Normál eloszlású napi hőmérséklet

# Szándékosan hozzáadunk anomáliákat (például extrém magas és alacsony hőmérsékletek)
temperatures[50] = 35  # Extrém magas hőmérséklet
temperatures[200] = -10  # Extrém alacsony hőmérséklet

# Hozzuk létre az idősor DataFrame-t
data = pd.DataFrame({'Temperature': temperatures}, index=pd.date_range(start="2024-01-01", periods=days, freq='D'))
```

```python
# **Isolation Forest**
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# Isolation Forest modell alkalmazása
model_if = IsolationForest(contamination=0.05, random_state=42)
model_if.fit(scaled_data)
predictions_if = model_if.predict(scaled_data)
data['Anomaly_IF'] = predictions_if
```

```python
# **Z-score alapú anomáliák**
z_scores = zscore(data['Temperature'])
threshold = 3  # Z-score > 3 esetén anomáliát keresünk
data['Anomaly_Z'] = np.where(np.abs(z_scores) > threshold, -1, 1)
```

```python
# Vizualizáljuk az anomáliákat
plt.figure(figsize=(14, 8))

# 1. ábra – Hőmérséklet adatok és Isolation Forest anomáliák
plt.subplot(3, 1, 1)
plt.plot(data.index, data['Temperature'], label='Hőmérséklet', color='blue')
plt.scatter(data.index[data['Anomaly_IF'] == -1],
            data['Temperature'][data['Anomaly_IF'] == -1],
            color='red', label='Isolation Forest anomáliák')
plt.title('Hőmérséklet – Isolation Forest által detektált anomáliák')
plt.legend()

# 2. ábra – Hőmérséklet adatok és Z-score anomáliák
plt.subplot(3, 1, 2)
plt.plot(data.index, data['Temperature'], label='Hőmérséklet', color='blue')
plt.scatter(data.index[data['Anomaly_Z'] == -1],
            data['Temperature'][data['Anomaly_Z'] == -1],
            color='green', label='Z-score anomáliák')
plt.title('Hőmérséklet – Z-score módszerrel detektált anomáliák')
plt.legend()

# 3. ábra – Összehasonlítás
plt.subplot(3, 1, 3)
plt.plot(data.index, data['Temperature'], label='Hőmérséklet', color='blue')
plt.scatter(data.index[data['Anomaly_IF'] == -1],
            data['Temperature'][data['Anomaly_IF'] == -1],
            color='red', label='Isolation Forest anomáliák')
plt.scatter(data.index[data['Anomaly_Z'] == -1],
            data['Temperature'][data['Anomaly_Z'] == -1],
            color='green', label='Z-score anomáliák')
plt.title('Isolation Forest és Z-score módszerek összehasonlítása')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Kiíratjuk az anomáliákat
if_anomalies = data[data['Anomaly_IF'] == -1]
zscore_anomalies = data[data['Anomaly_Z'] == -1]

print("Isolation Forest Anomalies:")
print(if_anomalies[['Temperature']])
print("\nZ-score Anomalies:")
print(zscore_anomalies[['Temperature']])
```

# Autoencoder - Anomáliák Detektálása

## Mi az Autoencoder?
Az **Autoencoder** egy olyan típusú neurális hálózat, amely az adatokat tömöríti és rekonstruálja. A modell célja, hogy az adatokat a lehető legjobban rekonstruálja, és a rekonstrukciós hibát minimalizálja. Az anomáliák detektálása során azokat az adatpontokat tekinthetjük anomáliának, amelyek nem rekonstruálhatók jól, mert eltérnek a normális adatokat képviselő mintáktól.

## Hogyan működik az Autoencoder?
1. **Encoder**: Az adatokat egy kisebb dimenziójú térbe tömöríti.
2. **Decoder**: A tömörített adatokat rekonstruálja.
3. **Rekonstrukciós hiba**: Az anomáliák akkor kerülnek felismerésre, amikor a rekonstruált adatpontok és az eredeti adatok közötti hiba nagy.



## Alkalmazás: Magasság, Testsúly és Életkor
Ebben az alkalmazásban három jellemzőt (magasság, testsúly, életkor) használtunk. Az **Autoencoder** segítségével detektáltuk az anomáliákat a rekonstrukciós hiba alapján.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam

# Generáljunk egy egyszerű adatot 3 paraméterrel: magasság (cm), testsúly (kg), életkor (év)
np.random.seed(42)
heights = np.random.normal(170, 10, 100)  # Magasság (170 cm átlag, 10 cm szórás)
weights = np.random.normal(70, 15, 100)   # Testsúly (70 kg átlag, 15 kg szórás)
ages = np.random.normal(30, 5, 100)       # Életkor (30 év átlag, 5 év szórás)

# Szándékosan hozzáadunk anomáliákat (például extrém magas és alacsony hőmérsékletek)
heights = np.append(heights, [250, 55])  # Extrém magasságok
weights = np.append(weights, [200, 30])  # Extrém testsúlyok
ages = np.append(ages, [5, 85])          # Extrém életkorok

# Hozzuk létre az idősor DataFrame-t
data = pd.DataFrame({'Height': heights, 'Weight': weights, 'Age': ages})

```

```python
# Skálázás
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# Autoencoder modell építése
model = Sequential()
model.add(Dense(64, activation='relu', input_dim=scaled_data.shape[1]))  # Bemeneti réteg
model.add(Dense(32, activation='relu'))  # Középső réteg
model.add(Dense(64, activation='relu'))  # Középső réteg visszafele
model.add(Dense(scaled_data.shape[1], activation='linear'))  # Kimeneti réteg

model.compile(optimizer=Adam(), loss='mean_squared_error')

# Modell tanítása
model.fit(scaled_data, scaled_data, epochs=50, batch_size=8, verbose=0)

```

```python

# Rekonstrukciós hiba kiszámítása
reconstructed_data = model.predict(scaled_data)
reconstruction_error = np.mean(np.abs(scaled_data - reconstructed_data), axis=1)
```

```python

# Anomáliák detektálása (pl. rekonstrukciós hiba > 2 szórás)
threshold = np.mean(reconstruction_error) + 2 * np.std(reconstruction_error)
anomalies = data[reconstruction_error > threshold]
```

```python
# Vizualizáció
plt.figure(figsize=(14, 8))

# Plot: Hőmérséklet adatok és rekonstrukciós hibák
plt.subplot(2, 1, 1)
plt.plot(data.index, reconstruction_error, label='Rekonstrukciós hiba', color='blue')
plt.axhline(y=threshold, color='red', linestyle='--', label='Hiba küszöb')
plt.title('Autoencoder - Rekonstrukciós Hiba')
plt.legend()

# Plot: Anomáliák kiemelése
plt.subplot(2, 1, 2)
plt.scatter(data.index, data['Height'], label='Data - Magasság', color='blue')
plt.scatter(anomalies.index, anomalies['Height'], color='red', label='Anomáliák')
plt.title('Autoencoder - Anomáliák detektálása (Magasság)')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Kiíratjuk az anomáliákat
print(f"Anomáliák:\n{anomalies[['Height', 'Weight', 'Age']]}")
```

# Autoencoder Idősoros Anomáliák Detektálása és Hasonlóság Mérés a Bottleneck Rétegen

## Mi az Autoencoder és hogyan használjuk idősoros anomáliákhoz?
Az **Autoencoder** egy típusú neurális hálózat, amely az adatokat tömöríti és visszaállítja. Az anomáliák detektálásakor azokat az adatpontokat tekintjük anomáliáknak, amelyek nem rekonstruálhatóak jól, mert eltérnek a normális adatokat képviselő mintáktól. Az **Autoencoder** alkalmazása idősoros adatokra különösen hasznos, mivel képes megtanulni a normális viselkedést, és azokat az adatokat jelzi, amelyek különböznek attól.

## Bottleneck Réteg és Hasonlóság Mérés
A **bottleneck réteg** az autoencoder alacsony dimenziós reprezentációját tartalmazza. A bottleneck rétegen alapuló **hasonlóság mérése** segíthet az adatok közötti hasonlóságok gyors meghatározásában. Azokat az adatpontokat, amelyek bottleneck reprezentációja jelentősen eltér, anomáliáknak tekinthetjük.

## Alkalmazás: Hőmérséklet Idősor
Ebben a példában három jellemzőt (magasság, testsúly, életkor) használtunk. Az **Autoencoder** segítségével detektáltuk az anomáliákat, és a bottleneck réteghez tartozó hasonlóságot használtuk az adatok közötti kapcsolat vizsgálatára.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from keras.models import Model, Sequential
from keras.layers import Dense, LSTM, Input
from keras.optimizers import Adam
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity

# Generáljunk egy szintetikus idősor adatot (pl. napi hőmérséklet)
np.random.seed(42)
days = 365  # 1 év
temperatures = np.random.normal(20, 5, days)  # Normál eloszlású napi hőmérséklet

# Szándékosan hozzáadunk anomáliákat (például extrém magas és alacsony hőmérsékletek)
temperatures[50] = 35  # Extrém magas hőmérséklet
temperatures[200] = -10  # Extrém alacsony hőmérséklet

# Hozzuk létre az idősor DataFrame-t
data = pd.DataFrame({'Temperature': temperatures}, index=pd.date_range(start="2024-01-01", periods=days, freq='D'))
```

```python

# Skálázás (Autoencoder érzékeny a skálázásra)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# Autoencoder modell építése
input_layer = Input(shape=(scaled_data.shape[1],))
encoded = Dense(64, activation='relu')(input_layer)
bottleneck = Dense(32, activation='relu')(encoded)  # Bottleneck réteg (alacsony dimenzió)
decoded = Dense(64, activation='relu')(bottleneck)
output_layer = Dense(scaled_data.shape[1], activation='linear')(decoded)

autoencoder = Model(input_layer, output_layer)
autoencoder.compile(optimizer=Adam(), loss='mean_squared_error')

```

```python

# Modell tanítása
autoencoder.fit(scaled_data, scaled_data, epochs=50, batch_size=8, verbose=0)

# Rekonstrukciós hiba kiszámítása
reconstructed_data = autoencoder.predict(scaled_data)
reconstruction_error = np.mean(np.abs(scaled_data - reconstructed_data), axis=1)

# Bottleneck réteg reprezentációja
encoder = Model(input_layer, bottleneck)
bottleneck_representations = encoder.predict(scaled_data)
```

```python
# Hasonlóságok mérése a bottleneck rétegen
similarity_matrix = cosine_similarity(bottleneck_representations)
```

```python

# Az anomáliák detektálása (rekonstrukciós hiba > küszöb)
threshold = np.mean(reconstruction_error) + 2 * np.std(reconstruction_error)
anomalies = data[reconstruction_error > threshold]
```

```python

# Vizualizáljuk az anomáliák hibáját és a hasonlóságokat
plt.figure(figsize=(14, 8))

# Plot: Rekonstrukciós hiba
plt.subplot(3, 1, 1)
plt.plot(data.index, reconstruction_error, label='Rekonstrukciós hiba', color='blue')
plt.axhline(y=threshold, color='red', linestyle='--', label='Hiba küszöb')
plt.title('Autoencoder - Rekonstrukciós Hiba')
plt.legend()

# Plot: Anomáliák detektálása
plt.subplot(3, 1, 2)
plt.scatter(data.index, data['Temperature'], label='Data - Hőmérséklet', color='blue')
plt.scatter(anomalies.index, anomalies['Temperature'], color='red', label='Anomáliák')
plt.title('Autoencoder - Anomáliák Detektálása')
plt.legend()

# Plot: Bottleneck hasonlóság
plt.subplot(3, 1, 3)
plt.imshow(similarity_matrix, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.title('Bottleneck Réteg Hasonlóság Mátrixa')
plt.tight_layout()
plt.show()
```

```python

# Kiíratjuk az anomáliákat
print(f"Anomáliák:\n{anomalies[['Temperature']]}")
```
